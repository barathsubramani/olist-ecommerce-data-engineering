```python
# DeltaLake_Optimization

sales_gold = spark.table(
    "workspace.olist_ecommerce.gold_sales_performance"
)

print("Gold sales records:", sales_gold.count())
print("Gold sales columns:", len(sales_gold.columns))

display(sales_gold.limit(10))
```

```python
detail = spark.sql("""
DESCRIBE DETAIL workspace.olist_ecommerce.gold_sales_performance
""")

display(detail)
```

```python
history = spark.sql("""
DESCRIBE HISTORY workspace.olist_ecommerce.gold_sales_performance
""")

display(history)
```

```python
history_df = spark.sql("""
DESCRIBE HISTORY workspace.olist_ecommerce.gold_sales_performance
""")

display(
    history_df.select(
        "version",
        "timestamp",
        "operation"
    )
)
```

```python
sales_v0 = (
    spark.read
    .format("delta")
    .option("versionAsOf", 0)
    .table("workspace.olist_ecommerce.gold_sales_performance")
)

print("Version 0 records:", sales_v0.count())
```

```python
spark.sql("""
OPTIMIZE workspace.olist_ecommerce.gold_sales_performance
""")

print("Delta optimization completed successfully!")
```
